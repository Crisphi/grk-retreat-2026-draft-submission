    Manage sleeptimes and retries of the crawling process to avoid getting blocked by YouTube. Uses an exponential sleep time duration. Generate a metadata csv with all relevant metadata per video from info.json

```python
"""
---------
Author: Cristian Ortega Singer
18-02-2026
---------
"""
#---- Imported Libraries
import csv
import os
import json

#---- Variables

# Setting directory paths

source_path = "./downloads/"
dest_file_path = "./downloads/master.csv"

# List to hold single entries for csv output
output = []

#---- Functions

# Fieldnames for CSV output
csv_fieldnames = [
    "id",
    "title",
    "description",
    "channel_id",
    "channel_url",
    "duration",
    "view_count",
    "age_limit",
    "webpage_url",
    "categories",
    "tags",
    "live_status",
    "media_type",
    "comment_count",
    "like_count",
    "channel",
    "channel_follower_count",
    "upload_date",
    "timestamp",
    "fulltitle",
    "duration_string",
    "was_live",
    "epoch",
    "format",
    "format_id",
    "language",
    "width",
    "height",
    "resolution",
    "fps",
    "aspect_ratio",
    "info_json_file",
    "description_file",
    "json3_file",
    "thumbnail_file",
    "video_file"
]


def extract_fields(json_data):
    fieldnames = csv_fieldnames
    row = {}

    for field in fieldnames:
        value = json_data.get(field, "$NONE$")

        # Convert lists to pipe-separated strings
        if isinstance(value, list):
            value = "|".join(str(v) for v in value)
        
        row[field] = value
    
    return row


def get_metadata():
    for root, dirs, files in os.walk(source_path):

        # Check if this directory contains an info.json
        info_files = [f for f in files if f.endswith(".info.json")]

        if not info_files:
            continue  # skip folders without video metadata

        # There should normally be exactly one per folder
        info_json = info_files[0]
        info_path = os.path.join(root, info_json)

        # Default sibling files
        description = "$NONE$"
        json3 = "$NONE$"
        thumbnail = "$NONE$"
        video = "$NONE$"

        # Detect siblings in same folder
        for f in files:
            if f.endswith(".description"):
                description = f
            elif f.endswith(".json3"):
                json3 = f
            elif f.endswith(".jpg") or f.endswith(".png") or f.endswith(".webp"):
                thumbnail = f
            elif f.endswith(".mp4") or f.endswith(".mkv") or f.endswith(".webm"):
                video = f

        try:
            with open(info_path, "r", encoding="utf-8") as fh:
                data = json.load(fh)
                entry = extract_fields(data)

            # Add sibling info
            entry["info_json_file"] = info_json
            entry["description_file"] = description
            entry["json3_file"] = json3
            entry["thumbnail_file"] = thumbnail
            entry["video_file"] = video

            output.append(entry)

        except Exception as e:
            print(f"Error reading {info_path}: {e}")

                
def write_csv():
    print("write csv")
    with open(dest_file_path, "w", newline="", encoding="utf-8") as csvfile:
        writer = csv.DictWriter(csvfile, fieldnames = csv_fieldnames)
        writer.writeheader()
        writer.writerows(output)

#---- Execution

get_metadata()
write_csv()
```

Search for duplicate values

```python
"""
---------
Author: Cristian Ortega Singer
18-02-2026
---------
"""
#---- Imported Libraries
import csv

#---- Variables

# Setting directory paths
csv_file_path = "./downloads/master.csv"

master_data = []

#---- Functions

#-- Load CSV Data
def load_csv():
    with open(csv_file_path, encoding="utf8", mode ='r') as file:    
       csv_reader = csv.DictReader(file)
       data = [row for row in csv_reader]
       return data

def search_for_duplicates(fieldname="id"):
    unique = []
    duplicates = []

    for row in master_data:
        val = row[fieldname]

        if val in unique:
            duplicates.append(val)
        else:
            unique.append(val)
    
    return unique, duplicates

#---- Execution

master_data = load_csv()
unique, duplicates = search_for_duplicates("title")

print("Duplicates:")
print(len(duplicates))
print(duplicates)
print("Unique:")
print(len(unique))
```
